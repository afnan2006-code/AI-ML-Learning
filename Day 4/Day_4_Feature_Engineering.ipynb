# Day 4 - Feature Engineering Pipeline

## Objective

The objective of this task is to transform the cleaned dataset from Day 3
into a machine-learning-ready numerical feature dataset.

The workflow includes:

- Datetime feature extraction
- Month extraction
- Day of Week extraction
- Weekend identification
- Categorical feature detection
- One-Hot Encoding
- Avoiding the Dummy Variable Trap

```python
import pandas as pd
import numpy as np
```

## 1. Load the Cleaned Dataset

The cleaned dataset generated during Day 3 is loaded as the starting point
for the feature engineering pipeline.

```python
df = pd.read_csv("../Day 3/final_clean_v2.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Dataset columns:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
# Check for datetime columns
datetime_columns = df.select_dtypes(
    include=["datetime64[ns]", "datetime64[ns, UTC]"]
).columns

print("Datetime columns:")
print(datetime_columns.tolist())
```

```python
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

## 2. Datetime and Categorical Feature Analysis

The Day 3 cleaned dataset was inspected for datetime and categorical
columns.

No datetime column is present in the dataset, so datetime features
such as Month, DayOfWeek, and Is_Weekend cannot be extracted without
introducing artificial data.

No categorical string columns are present either. The categorical
features from the original dataset have already been converted into
numerical encoded columns.

Therefore, no artificial datetime or categorical values are introduced.

```python
df_encoded = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=True
)

print("One-Hot Encoding completed.")
print("Shape:", df_encoded.shape)
```

```python
print("Final data types:")
print(df_encoded.dtypes)
```

```python
print("Final dataset shape:", df_encoded.shape)
```

```python
df_encoded.head()
```

```python
df_encoded.to_csv(
    "model_ready_features.csv",
    index=False
)

print("model_ready_features.csv created successfully!")
```

```python
final_df = pd.read_csv("model_ready_features.csv")

print("Final file shape:", final_df.shape)
print("Remaining categorical columns:")

print(
    final_df.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()
)
```

```python
print("Final dataset shape:", df_encoded.shape)

print("\nData types:")
print(df_encoded.dtypes)

print("\nMissing values:")
print(df_encoded.isnull().sum().sum())

print("\nRemaining categorical columns:")
print(
    df_encoded.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()
)
```

```python
df_encoded.to_csv(
    "model_ready_features.csv",
    index=False
)

print("model_ready_features.csv saved successfully!")
```
